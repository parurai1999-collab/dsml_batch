**Question (A)  Student** **Result** **Management** **System**


Create a user-defined function called student_result() that:

1.Takes the names and marks of 5 students as input.

2.Stores student data in a dictionary.

3.Uses conditional statements to assign grades:

A: 80 and above

B: 60–79

C: 40–59

F: Below 40

4.Displays each student's name, marks, and grade.
5.Calculates and displays:

Highest marks

Lowest marks

Average marks

```python
def student_result():
    students = {}
    marks_list = []

    # input
    for i in range(5):
        name = input("Enter name: ")
        marks = int(input("Enter marks: "))
        students[name] = marks
        marks_list.append(marks)
    for name in students:
        marks = students[name]

        if marks >= 80:
            grade = 'A'
        elif marks >= 60:
            grade = 'B'
        elif marks >= 40:
            grade = 'C'
        else:
            grade = 'F'

        print(name, "-", marks, "-", grade)

    print("Highest:", max(marks_list))
    print("Lowest:", min(marks_list))
    print("Average:", sum(marks_list) / 5)


student_result()
```

**Question** (**B**) **Smart** **Shopping** **Cart** **System**

Create a user-defined function called shopping_cart() that:

1.Allows the user to enter 5 product prices.

2.Stores the prices in a list.

3.Uses a loop to calculate the total bill.

4.Applies discounts using conditional statements:

Total ≥ 5000 → 15% discount

Total ≥ 3000 → 10% discount

Total ≥ 1000 → 5% discount

Otherwise → No discount

5.Displays:

Product list

Total amount

Discount amount

Final payable amount

```python
def shopping_cart():
    product_prices = []

    for i in range(5):
        prices = int(input("Enter the price: "))
        product_prices.append(prices)

    total = 0
    for p in product_prices:
        total += p

    if total >= 5000:
        discount = total * 0.15
    elif total >= 3000:
        discount = total * 0.10
    elif total >= 1000:
        discount = total * 0.05
    else:
        discount = 0

    final_amount = total - discount

    print("\nProducts:", product_prices)
    print("Total amount:", total)
    print("Discount amount:", discount)
    print("Final payable amount:", final_amount)


shopping_cart()
```
