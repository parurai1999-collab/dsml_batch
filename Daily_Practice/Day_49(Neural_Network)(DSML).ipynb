**What** **Is** **a** **Neural** **Network**?

A neural network is a set of simple computing units (neurons) arranged in layers,
connected by weighted links, that learns a mapping from input to output by
adjusting those weights from data.


**Biological** **vs** **Artificial** **Neuron**





**Anatomy** **of** **an** **Artificial** **Neuron**

Inputs (x)
The features you feed in - one number per
measurement.

Weights (w)
How much each input matters. Learned, not chosen.
Negative = works against the output.

Bias (b)
A shif't that lets the neuron fire even when all inputs
are zero,

Activation f

Squashes z into a useful range and adds non-linearity.

Output (9)
The prediction of this neuron - and the input to the
next layer.


**Activation** **Functions**

Without an activation function a hundred stacked layers collapse into one straight line - the network could never learn a curve, a boundary or an
image.



Sigmoid

o(z) = 1 /(1+e"2)

Squashes to [0, 1). Reads as a
probability.

Use: Output layer, binary classification




Tanh

tanh(z)

Squashes to (-1, 1), zero-centred.

Use: Hidden layers in older / recurrent



ReLU

max(0,z)

Passes positives, blocks negatives.
Fast, no vanishing gradient.

Use: Default for hidden layers in deep
nets


Softmax

Turns scores into probabilities
summing to 1.


Use: Output layer, multi-class
classification

Every neuron in a layer connects to every neuron in the next - a fully connected (dense) network.


What each layer does

Input layer

One neuron per feature. No computation - it
just holds the data.

Hidden layers

Where the learning lives. Each layer builds
more abstract combinations of the previous
one.

Output layer

Shape depends on the task: 1 neuron + sigmoid
for yes/no, N neurons + saftmax for N classes,
1 linear neuron for a number.

Depth

"Deep learning" simply means more than one
hidden layer.

**Two** **Directions** **of** **Flow**

**FEEDFORWARD**

Making the initial prediction

Data travels input -> hidden -> output. Each neuron computes a weighted
sum, applies its activation and passes the result on. Nothing is learned
here - the network simply answers with the weights it currently has.

**BACKPROPAGATION**

Learning from mistakes

The error at the output travels back output -> hidden -> input. Each
weight is told how much it contributed to the mistake, and is nudged in
the direction that reduces it.

**Measuring** **the** **Mistake**: **Loss** **Functions**

The loss function turns "how wrong was the network" into a single number. Training is nothing more than making that number smaller.


**Mean** **Squared** **Error** (MSE)

L=1/nE(y-ý)2

**Binary** **Cross**-**Entropy**


L =- [y.logŷ+(1-y)-log(1-ŷ)]

Binary classification - yes / no

e.g. Spam or not, pass or fail




**Categorical** **Cross**-**Entropy**


L =- Eyi . log ŷi

Multi-class classification

e.g. Digit 0-9, cat / dog / bird

**Types** **of** **Neural** **Networks**

**Feedforward** **Neural** **Network**

Straight line of layers, no loops. The general-purpose
baseline for tabular data.

Best for: tabular data & simple prediction



**Convolutional** **Neural** **Network**

Filters slide over a grid to detect local patterns. Built
for images.

Best for: images & video



**Recurrent** **Neural** **Network**


Has memory - output feeds back in. Built for
sequences and time.

Best for: text, speech & time series


**Generative** **Adversarial** **Network**

Two networks compete: one generates, one judges.
Built to create.

Best for: image / data generation

```python
import torch
import torch.nn as nn
```

```python

# 1. Define the Feed-Forward Neural Network class
class FeedForwardNN(nn.Module):
    def __init__(self, input_size, hidden_size, num_classes):
        # Call the constructor of the parent nn.Module class
        super(FeedForwardNN, self).__init__()

        # First fully connected layer (Input -> Hidden)
        self.fc1 = nn.Linear(input_size, hidden_size)

        # Activation function
        self.relu = nn.ReLU()

        # Second fully connected layer (Hidden -> Output)
        self.fc2 = nn.Linear(hidden_size, num_classes)

    def forward(self, x):
        # Pass input through the first linear layer
        out = self.fc1(x)

        # Apply the ReLU activation function
        out = self.relu(out)

        # Pass through the output layer
        out = self.fc2(out)

        return out

```

```python
# 2. Instantiate the model with specific dimensions
INPUT_DIM = 784    # Example: Flattened 28x28 image pixels
HIDDEN_DIM = 128   # Number of units in the hidden layer
OUTPUT_DIM = 10    # Example: 10 target classes (0-9 digits)

model = FeedForwardNN(input_size=INPUT_DIM, hidden_size=HIDDEN_DIM, num_classes=OUTPUT_DIM)
print("Model Architecture:\n", model)

# 3. Simulate a forward pass with dummy data
# Batch size = 4, Input features = 784
dummy_input = torch.randn(4, INPUT_DIM)
predictions = model(dummy_input)

print("\nOutput Shape (Batch Size, Classes):", predictions.shape)
```

```python
import tensorflow as tf
mnist = tf.keras.datasets.mnist

(x_train, y_train),(x_test, y_test) = mnist.load_data()
x_train, x_test = x_train / 255.0, x_test / 255.0

model = tf.keras.models.Sequential([
  tf.keras.layers.Flatten(input_shape=(28, 28)),
  tf.keras.layers.Dense(128, activation='relu'),
  tf.keras.layers.Dropout(0.2),
  tf.keras.layers.Dense(10, activation='softmax')
])

model.compile(optimizer='adam',
  loss='sparse_categorical_crossentropy',
  metrics=['accuracy'])

model.fit(x_train, y_train, epochs=5)
model.evaluate(x_test, y_test)
```
