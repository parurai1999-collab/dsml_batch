```python
Mini Project: Student Management System

Project Title:

Student Management System Using OOP Concepts

Problem Statement:

Develop a Student Management System using Python that demonstrates
the concepts of Class, Object, Encapsulation, Inheritance, and Abstraction.
The system should allow users to store student information, calculate
grades, and display student details based on different student categories.

Requirements

1. Create an Abstract Class Student

The abstract class should contain:
Student ID
Student Name
A private attribute__marks
A method to set marks
A method to get marks
An abstract method calculate_grade()
A method display_details()
```

```python
from abc import ABC, abstractmethod
class Student(ABC):

    def __init__(self, student_id, name):
        self.student_id = student_id
        self.name = name
        self.__marks = 0

    def set_marks(self, marks):
        if 0 <= marks <= 100:
            self.__marks = marks
        else:
            print("Invalid marks")

    def get_marks(self):
        return self.__marks

    @abstractmethod
    def calculate_grade(self):
        pass

    def display_detail(self):
        print("Student ID :", self.student_id)
        print("Name       :", self.name)
        print("Marks      :", self.get_marks())


class RegularStudent(Student):

    def calculate_grade(self):

        marks = self.get_marks()

        if marks >= 80:
            return "A"

        elif marks >= 60:
            return "B"

        elif marks >= 40:
            return "C"

        else:
            return "Fail"


student1 = RegularStudent(101, "Ram")

student1.set_marks(78)

student1.display_detail()
```
