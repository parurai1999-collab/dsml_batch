```python
import pandas as pd
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
```

```python
# Load Dataset
df = sns.load_dataset("titanic")
```

```python
df.columns
```

```python
# Select numerical features
X = df[["pclass", "age", "parch", "fare"]]
```

```python
# Fill missing values
X = X.fillna(X.mean())
```

```python
# standardize the data
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
X_scaled
```

```python
# Appy PCA
pca = PCA(n_components = 2)
x_pca = pca.fit_transform(X_scaled)
```

```python
pca_df = pd.DataFrame(x_pca, columns=["PC1", "PC2"])
pca_df.head()
```

LDA stands for Linear Discriminant Analysis

1) LDA is a supervised machine learning algorithm primarily used to dimessionality reduction and classification taks
2) the goal is to find a linear combination of features that maximizes the seperation of class while minimizing the spread (variance) within  each class
 - maximizing mean difference
 - minimizing within class scatter


image.png

```python
# 1) LDA is supervised machine learning algorithm primarily used to dimensionality reduction and classification tasks
# 2) the goal is to find a linear combination of features that maximizes the seperation of class while minimizing the spread(variance) within each class
# -maximize mean difference
# -minimizing variance
```

```python
from numpy import matrix
# Steps for LDA
1) Calculate the mean vectors for each class
2) Compute the covariance of each class
3) compute the within class scatter matrix
4) compute the between class scatter matrix
5) find eigen values and choose top eigen value
6) find the eigen vectors
7) find the dot product of the data points with eigen vector
```

```python
# Applying LDA
# Target variable
y = df["survived"]
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
lda = LinearDiscriminantAnalysis(n_components = 1)
X_lda = lda.fit_transform(X_scaled, y)
```

```python
lda_df = pd.DataFrame(X_lda, columns = ["LD1"])
lda_df.head()
```

Explained variance ratio is a measure of the percentage of the total variance in a given dataset for each principal component. The explained variance ratio of a principal component is measured as the ratio of its eigenvalue to the sum of the eigenvalues of all the principal components

```python

print("Explained Variance Ratio:")
print(pca.explained_variance_ratio_)

print("\nTotal Variance Explained:")
print(sum(pca.explained_variance_ratio_))
```
