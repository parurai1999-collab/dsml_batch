**Duck** **Type**

Python doesn't care about the
object's class; it only cares
whether the object has the
required method

```python
class Duck:
  def swim(self):
    return "Duck is swimming."
  def fly(self):
    return "Duck flying"

class Airplane:
  def fly(self):
    return "Airplane flying"

def fly_test(entity):
  print(entity.fly())

duck = Duck()
airplane = Airplane()
fly_test(duck)
fly_test(airplane)
```

**Class** **Task**

Create classes Esewa, Khalti, and
BankTransfer, each having a pay(amount)
method. Write a payment processor function
that works with any payment method object

```python
class Esewa:
  def pay(self, amount):
    return f"Esewa paying {amount}"
class Khalti:
  def pay (self, amount):
    return f"Khalti paying {amount}"

class BankTransfer:
  def pay (self, amount):
    return f"Bank Transfer paying {amount}"
def payment_processor(entity,amount):
  print(entity.pay(amount))
esa = Esewa()
khal = Khalti()
bank = BankTransfer()
payment_processor(esa, 100)
payment_processor(khal, 200)
payment_processor(bank, 300)

```

**Magic** **Method**

magic methods (also called dunder methods, short for
"double underscore") are special methods with names
surrounded by double underscores, such as __init__,
__str__, and __len__.



These given below are called magic method.

```python
dir(Duck)
```

```python
class BankCustomer:
  def __init__(self, name, age):
   self.name = name
   self.age = age
  def __str__(self):
    return f"{self.name} is {self.age} years old"
  def __repr__(self):
    return f"BankCustomer ('{self.name}', '{self.age}')"
  def __len__(self):
    return self.age
  def __call__(self, gender = "Male"):
   return f"{self.name} is a {gender}"

c = BankCustomer("Ram", 25)
print(c)
print(repr(c))
print(len(c))
print(c("Female"))
```

**Class** **Task**

Create a BankAccount class and implement __str__()
and __repr__() methods to display account details in
user-friendly and developer-friendly formats.

```python
class BankAccount:
  def __init__(self, name, balance):
    self.name = name
    self.balance = balance
  def __str__(self):
    return f"{self.name} account balance is {self.balance}"
  def __repr__(self):
    return f"BankAccount ('{self.name}', '{self.balance}')"
c1 = BankAccount("Paru", 200000)
print(c1)
print(repr(c1))
print(str(c1))
```

str(obj) method is used to return a string representation of an object that is
meant to be human-readable. str is human readable form.

repr(obj) method, on the other hand, is used to return a string representation of an object that is meant to be unambiguous and suitable for debugging.
