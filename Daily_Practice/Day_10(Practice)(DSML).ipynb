**Write** **a** **function** **that** **takes** **three** **numbers**
**as** **arguments** **and** **returns** **the** **largest**
**number**.

```python
a = int(input("Enter the first number:"))
b = int(input("Enter the second number:"))
c = int(input("Enter the third number:"))
def largest_number(a,b,c):
  return max(a,b,c)
print("The largest number is", largest_number(a,b,c))
```

**Write** **a** **function** **that** **takes** **a** **string** **and**
**returns** **the** **number** **of** **vowels** (**a**, **e**, **i**, **o**, **u**)
**present** **in** **it**.

```python
strng = input("Enter a string:")
vowel_letter = ["a", "e", "i", "o", "u"]
def vowel(strng):
  count = 0
  for char in strng:
    if char in vowel_letter:
      count += 1

  return count
print(vowel(strng))
```

**open**() **build**-**in** **function** **in** **python**

```python
file = open("simple.txt", mode="r") # open file and close file
print(file)
file.close()
```

```python
lst = []
def add(n):
  for i in range(n):
    a = int(input("Enter a number:"))
    lst.append(a)

  file = open("userData.txt", mode = "w")
  file.write(str(lst))
  file.close()
```

```python
add(57)
```

```python
with open("userData.txt", mode= "r") as file:
  print(file.read())
```

**enumerate**() **function** **in** **python**

```python
values = [1,2,3,4,5,6,7]
for i in enumerate(values, start=1):
  print(i)
```

```python
values1 = []
for i, values in enumerate(values, start=0):
  print(i, values1)
```

```python
dict = {
    "name" : "Paru",
    "gender": "female",
    "DOB" : 2026-04-3,
    "salary": 1000000
}
for i, (k, v) in enumerate(dict.items())
  print(k,v)
```
