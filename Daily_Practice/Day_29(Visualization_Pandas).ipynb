**Bar** **Graph**

```python
# Bar Graph
# Gender Count
import matplotlib.pyplot as plt
df["Sex"].value_counts().plot.bar()
plt.title("Gender Distribution")
plt.show()
```

```python
df["Survived"].value_counts().plot(
    kind = "bar",
    title = "Count of Survived Passenger",
    figsize=(5,5),
    legend = True,
    yticks = list(range(0,600,50)),
    xlabel = "Survived (0=No, 1=Yes)",
    ylabel = "No of Passengers",
    grid = True,
    colormap = "bwr"
)
```

```python
import matplotlib.pyplot as plt

pd.crosstab(df.Survived, df.Sex).plot(
    kind="bar",
    title="Count of Survived Passengers",
    figsize=(5,5),
    legend=True,
    ylabel="No of Passengers",
    grid=True,
    colormap="Dark2"
)

plt.xlabel("Survived (0 = No, 1 = Yes)")
plt.xticks(rotation=0)
plt.show()
```

**Pie** **Chart**

```python
import matplotlib.pyplot as plt

group_by_embarked = df.groupby("Embarked")["Age"].count()

group_by_embarked.plot(
    kind="pie",
    figsize=(5,5),
    autopct="%1.1f%%",
    title="Passengers by Embarked"
)

plt.ylabel("")
plt.show()
```

```python
import matplotlib.pyplot as plt

df["Survived"].value_counts().plot(
    kind="pie",
    figsize=(5,5),
    autopct="%1.1f%%",
    title="Survival Distribution (0 = No, 1 = Yes)",
    colormap="Dark2"
)

plt.ylabel("")
plt.show()
```

```python
df["Survived"].value_counts().plot.pie(
    autopct = %1.1f%%,
    figsize = (5,5),

)
```

```python
import matplotlib.pyplot as plt

pd.crosstab(df.Survived, df.Sex).plot(
    kind="pie",
    subplots=True,
    figsize=(6,8),
    fontsize=10,
    colormap="Dark2"
)

plt.show()
```

**Line** **Graph**

```python
# Line Graph
df["Age"].loc[:800].plot.line(figsize = (10,4))
```

```python
df[["Age", "Fare"]].loc[:800].plot.line(figsize=(10,4))
```

**Scatter Graph**

```python
df[["Age", "Fare"]].loc[:800].plot.scatter(
    x = "Age",
    y = "Fare",
    figsize = (10,4)
)
```

An outlier is a data point that is very different from the rest of the data

```python
import matplotlib.pyplot as plt

df_clean = df[["Age", "Fare"]].loc[:800].dropna()

df_clean.plot.scatter(
    x="Age",
    y="Fare",
    figsize=(10,4),
    alpha=0.5,
    color="blue"
)

plt.title("Age vs Fare")
plt.xlabel("Age")
plt.ylabel("Fare")
plt.grid(True)
plt.show()
```

**Histograph**

```python
# Histogram
import matplotlib.pyplot as plt
df["Age"].plot.hist(
    bins = 20,
    figsize = (6,4),
    color = "blue"
)
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.grid(True)
plt.show()
```

```python
df["Fare"].plot.hist(
    bins = 20,
    figsize = (6,4),
    color = "green"
)
plt.title("Fare Distribution")
plt.xlabel("Fare")
plt.ylabel("Frequency")
plt.grid(True)
plt.show()
```
