**Display** **all** **student** **names** **with** **their** **serial** **numbers** **starting** **from** **1**.

```python
students = ["Ram","Sita","Hari","Gita","Shyam","Rita"]
for i, value in enumerate(students, start=1):
    print(i, value)
```

**Combine** **the** **student** **names** **and** **marks** **into** **a** **single** **collection** **containing**
**student** **records**.

```python
students = ["Ram", "Sita", "Hari", "Gita", "Shyam", "Rita"]
marks = [85, 42, 78, 91, 35, 67]
student_records = dict(zip(students, marks))
print(student_records)
```

**Generate** **grades** **for** **each** **student** (**like** **in** **SEE**)

```python
students = ["Ram", "Sita", "Hari", "Gita", "Shyam", "Rita"]
marks = [85, 42, 78, 91, 35, 67]
def student_result(marks):

    if marks >= 90:
        return "A+"
    elif marks >= 80:
        return "A"
    elif marks >= 70:
        return "B+"
    elif marks >= 60:
        return "B"
    elif marks >= 50:
        return "C+"
    elif marks >= 40:
        return "C"
    else:
        return "F"


for key, value in student_records.items():
    grade = student_result(value)
    print(f"Name: {key} Marks: {value} Grade: {grade}")
```

**Create** **a** **list** **containing** **only** **the** **students** **who** **have** **passed** **the**
**examination**. (**marks** **above** **40** **is** **pass**)

```python
students = ["Ram", "Sita", "Hari", "Gita", "Shyam", "Rita"]
marks = [85, 42, 78, 91, 35, 67]
def passed_student(marks):
    return marks > 40


result = list(filter(lambda x: passed_student(x[1]), zip(students, marks)))
for name, marks in result:
    print(f"Name: {key} Marks: {marks}")
```

**Calculate** **the** **total** **marks** **obtained** **by** **all** **students** **combined**.


```python
from functools import reduce
students = ["Ram", "Sita", "Hari", "Gita", "Shyam", "Rita"]
marks = [85, 42, 78, 91, 35, 67]
def add_marks(x, y):
    return x+y


total_marks = reduce(add_marks, marks)
print("Total Marks:", total_marks)
```

**Find** **the** **highest** **mark** **among** **all** **students** **without** **using** **the** **built**-**in** **max**()
**function**

```python
from functools import reduce
students = ["Ram", "Sita", "Hari", "Gita", "Shyam", "Rita"]
marks = [85, 42, 78, 91, 35, 67]
def highest_mark(x, y):
    if x > y:
        return x
    else:
        return y


highest = reduce(highest_mark, marks)
print("Highest Mark:", highest)
```
