```python
class BankAccount:
    def __init__(self, name, balance):
        self.name = name
        self.__balance = balance   # private variable

        print("Account created for:", self.name)
        print("Initial balance:", self.__balance)

    def deposit(self):
        amount = float(input("Enter the amount to deposit: "))
        if amount > 0:
            self.__balance += amount
            print(f"New balance is {self.__balance}")
        else:
            print("Enter a valid amount")

    def withdraw(self):
        amount = float(input("Enter the amount to withdraw: "))
        if amount > self.__balance:
            print("Not enough balance")
        elif amount <= 0:
            print("Enter a valid amount")
        else:
            self.__balance -= amount
            print(f"New balance is {self.__balance}")

    def check_balance(self):
        print(f"Current balance is {self.__balance}")
```

```python
class SavingAccount(BankAccount):
  def add_interest(self, rate=5):
    balance = self.check_balance()
    interest = balance * (rate/100)
    print(f" The interest is {interest}")
    return interest
```

```python
print("--Enter the account details--")
name = input("Enter your name: ")
amount = int(input("Enter the amount: "))

obj = BankAccount(name, amount)

obj.deposit()
obj.withdraw()
obj.check_balance()
```

**Encapsulation** **in** **Python**

It is a way of hiding important details and protecting data from direct access.

```python
class Student:
    def __init__(self):
        self.name = "Paru"          # Public
        self._age = 20              # Protected
        self.__marks = 85           # Private

    def show_marks(self):
        return self.__marks   # Accessing private inside class


# Create object
s = Student()

# Public access ✅
print(s.name)     # Works

# Protected access (possible but not recommended)
print(s._age)

# Private access (will cause error)
# print(s.__marks)  # AttributeError

# Access private using method
print(s.show_marks())

```

**Abstraction** **in** **Python**

Abstraction means hiding internal details and showing only the important features to the user.


In simple words:
You use things without knowing how they actually work inside.


```python
from abc import ABC, abstractmethod

# Abstract class
class Animal(ABC):

    @abstractmethod
    def sound(self):
        pass   # no implementation


# Child class
class Dog(Animal):
    def sound(self):
        return "Bark"


# Child class
class Cat(Animal):
    def sound(self):
        return "Meow"


# Objects
d = Dog()
c = Cat()

print(d.sound())   # Bark
print(c.sound())   # Meow
```
