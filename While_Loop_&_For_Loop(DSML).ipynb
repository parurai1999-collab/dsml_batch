**Python While Loop in Python**


**Sum** **of** **10** **number**

```python
sum = 0
i = 1
while i <= 10:
  sum = sum + i
  i+=1
print(" Sum of ten number:", sum)
```

**Infinite Loop**

Forecasting data weather where data continuously flow. In case of game, until quit it will run.

```python
while True:
  print("Hello, world")
```

**Break statement**

```python
i = 1
while i <= 10:
  print(i)
  i+=1
  if i == 3:
    break
```

**Continue Statement**

```python
i = 0
while i < 10:
  i+=1
  if i == 3:
    continue
```

**Pass** **Statement**

We make a function but it doesnot need right now so we simply pass the function. To pass the function we will use pass statement. It is a placeholder.

**Range**(**start**, **end**, **step**)

```python
stored_value = 7
user_number = int(input("Enter the number:"))
for i in range(10):
  if user_number == stored_value:
   print("Guess correct")
   break
  else:
    print("wrong")
```
