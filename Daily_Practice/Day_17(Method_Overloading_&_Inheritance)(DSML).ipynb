In python functional overloading is not supported.
In that case we use default argument.

```python
class Calculator:
   def add(self, a,b,c):
    return a+b+c
```

```python
obj1 = Calculator()
obj1.add(2,3)
```

```python
obj2 = Calculator()
obj2.add(2,3,4)
```

```python
class Calculator:
   def add(self, a,b,c=0): #Default argument, we solve method overloading
    return a+b+c
```

```python
obj3 = Calculator()
obj3.add(2,3)
```

Another way for method loading is using *args and **kwargs

```python
class Calculator:
  def add(self, *args):
    return sum(args)
```

```python
obj4 = Calculator()
obj4.add(1)
```

```python
obj5 = Calculator()
obj5.add(1,2)
```

Multiple Inhertance

```python
class Animal:
  def sound(self):
    print("Animal makes sound")
class Dog():
  def bark(self):
    print("Dog barks")
class cat(Animal, Dog):
 def meow(self):
  print("Cat meows")
c1 = cat()
c1.bark()
c1.meow()
c1.sound()
```

Multilevel Inheritance

```python
class Animal:
  def sound(self):
    print("Animal makes sound")
class Dog(Animal):
  def bark(self):
    print("Dog barks")
class cat(Dog):
 def meow(self):
  print("Cat meows")
```

```python
c2 = Dog()
c2.bark()
c2.sound()
```

```python
class Person:
  def __init__(self, name, age):
    self.name = name
    self.age = age
  def student(self):
    print(f"Name: {self.name}, Age : {self.age}")
```

```python
class Person:
    def __init__(self, name, age, **kwargs):
        self.name = name
        self.age = age

    def show_person(self):
        print(f"Name: {self.name}, Age: {self.age}")


class Student(Person):
    def __init__(self, student_id, **kwargs):
        super().__init__(**kwargs)
        self.student_id = student_id

    def show_student(self):
        print(f"Student ID: {self.student_id}")


class Teacher(Person):
    def __init__(self, subject, **kwargs):
        super().__init__(**kwargs)
        self.subject = subject

    def show_teacher(self):
        print(f"Subject: {self.subject}")


class TeacherAssistant(Student, Teacher):
    def __init__(self, name, age, student_id, subject):
        super().__init__(
            name=name,
            age=age,
            student_id=student_id,
            subject=subject
        )

    def show_all(self):
        self.show_person()
        self.show_student()
        self.show_teacher()
        print("Role: Teacher Assistant")



ta = TeacherAssistant("Paru", 23, 101, "Math")
ta.show_all()
```
