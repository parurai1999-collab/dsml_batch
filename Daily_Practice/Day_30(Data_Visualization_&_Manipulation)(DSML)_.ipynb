**Central** **Tendency**

```python
# Central Tendency
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
```

```python
x =np.random.standard_t(500, size = 1000)
x[:15]
```

**Mean**

```python
_mean  =np.mean(x)
print(f"Mean of x:{_mean}")
```

```python
plt.hist(x,bins = 20, color= "blue")
```

```python
plt.axvline(x=_mean, color = 'yellow')
```

```python
plt.hist(x,bins = 20, color= "blue")
plt.axvline(x=_mean, color = 'yellow')
```

**Median**

```python
import matplotlib.pyplot as plt

_median = np.median(x)
print(f"Median of x: {_median}")

plt.hist(x, bins=20, color="red")
plt.show()
```

```python
plt.hist(x,bins = 20, color= "red")
```

```python
plt.axvline(x=_median, color = 'yellow')
```

```python
plt.hist(x,bins = 20, color= "red")
plt.axvline(x=_median, color = 'yellow')
```

**Mode**

```python
from scipy import stats

_mode = stats.mode(x).mode
print(f"Mode of x: {_mode}")
```

```python
plt.hist(x,bins = 20, color= "green")
```

```python
plt.axvline(x=_mode, color = 'yellow')
```

```python
plt.hist(x,bins = 20, color= "red")
plt.axvline(x=_mode, color = 'yellow')
```
