**Iterators** **in** **Python**

An iterable is an object that can be looped over.

An iterator is an object that:

Gives values one at a time
Uses next() internally
Remembers its current position

```python
num =[1,2,3]
it = iter(num)
```

```python
print(next(it))
```

```python
def square(n):
  for i in range(n):
    return i**2
```

```python
square(3)
```

**Generators** **in** **Python**

A generator is a special type of function that:

Returns values one at a time
Uses the keyword yield
Does not store all values in memory

✅ Generators are iterators.

```python
def square(n):
  for i in range(n):
    yield i**2
```

```python
a = square(3)
```

```python
print(next(a))
```

yield returns a value and pauses the function execution.

It resumes from the same point when next value is requested.

Generators using yield do not store all values in memory, making them memory‑efficient.


return = end function

yield = pause function

**Common** **Built**-**in** **Exceptions**

```python
try:
  number = [1,2,3]
  print(number[10])
except IndexError:
  print("Index out of range!")
finally:
  print("Closed")
```

```python
try:
  file = open("data.txt", "r")
  print(file)
except FileNotFoundError:
  print("File not found!")
finally:
  print("Closed")
```

**Raise** **an** **Exception**

```python
def check(num):
  if num < 0:
    raise ValueError("Number must be positive.")
  print("number:", num)
```

```python
check(-5)
```

```python
try:
  check(-5)
except ValueError:
  print("Error: Number must be positive")
```
