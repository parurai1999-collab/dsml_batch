```python
txt = "Hello! Python World"
print(txt.lower())
print(txt.upper())
print(txt.isalpha())
print(txt.isdigit())
print(txt.isalnum())
print(txt.capitalize())
print(txt.split())
print(txt.replace('Python', 'DSML'))
print(txt.find('Python'))
print(txt.index('Python'))
```

```python
sentence = input("Enter a sentence:")
print("UpperCase:", sentence.upper())
print("LowerCase:", sentence.lower())
print("Length:", len(sentence))
print("Indexing:", sentence.find('a')) # Changed to find() and provided an example argument
print("Checking1:", sentence.isalpha()) # Corrected from isalpa()
print("Checking2:", sentence.isalnum())
old_word = input("Enter the word to be replaced:")
new_word = input("Enter the new word:")
print("Replaced:", sentence.replace(old_word, new_word)) # Used variables instead of literal strings
```
