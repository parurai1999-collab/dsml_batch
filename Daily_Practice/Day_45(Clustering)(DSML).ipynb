**K-Means** Clustering is an unsupervised machine learning algorithm used to group similar data points into K clusters.


**“K” = number of clusters \
“Means” = average (centroid of cluster)**

It finds patterns in data without labels

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
```

```python
iris = load_iris()
X = iris.data
y = iris.target
```

```python
X
```

```python
y
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
X_scaled #normalization
```

```python
inertia = []
for k in range(1, 11): # Test k for 1 to 10
 kmeans = KMeans(n_clusters = k, random_state= 42)
 kmeans.fit(X_scaled)
 inertia.append(kmeans.inertia_) # sum of squared distances to cluster center
```

```python
plt.figure(figsize= (6,4))
plt.plot(range(1,11), inertia, marker = "o")
plt.xlabel("Number of cluster (k)")
plt.ylabel("Inertia")
plt.title("Elbow Method for Optional k")
plt.show()
```

```python
from sklearn.metrics import silhouette_score
```

```python
silhoutte_score = []
# The silhouette score cannot be calculated for a single cluster because it measures the distance between a point and its cluster compared to the nearest neighboring cluster. With only one cluster,
#  there are no neighboring clusters, making the silhouette formula mathematically undefined
for k in range(2,11):
  kmeans = KMeans(n_clusters=k, random_state=42)
  labels_temp = kmeans.fit_predict(X_scaled)
  score = silhouette_score(X_scaled, labels_temp)
  silhoutte_score.append(score)
```

```python
plt.figure(figsize=(6,4))
plt.plot(range(2,11), silhoutte_score, marker='o')
plt.xlabel("Number of cluster (k)")
plt.ylabel("Silhoutte Score")
plt.title('Silhoutte Method for Optimal k')
plt.show()
```

```python
best_k = range(2,11)[silhoutte_score.index(max(silhoutte_score))]
print("Best K Silhoutte:", best_k)
```

```python
#train final k means model
kmeans = KMeans(n_clusters=3,random_state=42)
kmeans.fit(X_scaled)
```

```python
labels = kmeans.labels_
centroids = kmeans.cluster_centers_
```

```python
labels
```

```python
plt.figure(figsize=(6,4))
plt.scatter(X_scaled[:,0], X_scaled[:,1], c=labels, cmap='viridis',s=50) #Datapoint
plt.scatter(kmeans.cluster_centers_[:,0],kmeans.cluster_centers_[:,1],
            c='red',marker='X',s=200, label='Centroid')
plt.xlabel(iris.feature_names[0])
plt.ylabel(iris.feature_names[1])
plt.title('K-mean Clustering on Iris Dataset')
plt.legend()
plt.show()
```
