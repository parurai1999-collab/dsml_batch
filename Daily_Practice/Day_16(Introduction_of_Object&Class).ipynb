**Attribute**

Types of Attributes
1. Instance Attributes
2. Class attributes


```python
class Student:
  '''Students Information'''
  school = "SkillShikshya" # Class attribute
  def __init__(self, name, age):
    self.name = name #instance attribute
    self.age = age
  @classmethod  # sabai tira use garna sakdimna
  def get_school(cls):
   print(cls.school)
Student.get_school()
s1 = Student("Paru", 22)
print(s1.name)
print(s1.age)
```

**Calculator**

```python
class Calculator:
  def __init__(self, num1, num2):
    self.num1 = num1
    self.num2 = num2
  def add(self):
    return self.num1 +self.num2
  def sub(self):
    return self.num1 - self.num2
c1 = Calculator(10, 20)
print(c1.add())
print(c1.sub())
```

```python
class Calculator:
  @staticmethod
  def add(a,b):
    return a+b
print(Calculator.add(4,5))
```

Types of Methods
1. Instance methods
2. Class methods
3. Static methods


```python
class Dog:
  count = 0 # class attribute
  def __init__(self, name, age=3):
    self.name = name
    self.age = age
    Dog.count += 1

  def bark(self):
    print(f"{self.name} is barkin")
  @classmethod
  def get_count(cls):
    return cls.count

  @staticmethod
  def get_vet_ingo():
    return "Dr. Ram K.C"
```

```python
dog1 = Dog("puppy")
print(dog1.name)
print(dog1.age)
```

```python
dog1.get_count() # kati choti initialize vako xa
```

```python
dog1.get_vet_ingo()
```
