**Deep** **Copy** **and** **Shallow** **copy**

```python
lst3 = [[1,2,3], [4,5,6]]
lst4 = lst3.copy()
lst3[1][1] = 100
print(lst3)
print(lst4)


```

```python
import copy
lst5 = [[1,2,3], [4,5,6]]
lst6 = copy.deepcopy(lst5)
lst5[1][2] = 100
print(lst5)
print(lst6)
```
