AI (Artificial Intelligence)

AI is the broad field of making machines “intelligent,” meaning they can perform tasks that usually require human intelligence.

Goal: To simulate human reasoning, learning, problem-solving, and decision-making.

Examples in real life:

Siri or Google Assistant

Self-driving cars

Recommendation systems (Netflix, YouTube)

AI doesn’t always learn; sometimes it just follows rules (like chess algorithms)

ML is a subset of AI where machines learn from data instead of being explicitly programmed rules.

Goal: Teach machines to identify patterns and make predictions based on data.

How it works:

You give the model data (e.g., images, text, numbers).
It learns patterns from the data.
It can predict or make decisions for new unseen data.

Examples:

Spam email detection

Fraud detection in banks

Stock price prediction

ML requires data to improve performance over time.


NN (Neural Network)

NN is a type of ML model inspired by the human brain. It consists of layers of “neurons” (nodes) that process data.

Goal: Learn complex patterns in data, especially when relationships are nonlinear.

Structure:

Input layer: receives data

Hidden layers: process data through mathematical operations

Output layer: gives prediction or classification

Examples: Image recognition (detect cats vs dogs)

Voice recognition (Alexa, Google Assistant)

Language translation (Google Translate)

Key point: Neural networks are very powerful for complex tasks like vision, speech, and
NLP

Linear Regression

Linear Regression is a fundamental supervised learning
algorithm used to model the relationship between a
dependent variable and one or more independent
variables. It predicts continuous values by fitting a
straight line that best represents the data.


Why Linear Regression Fails Here

Predictions are not constrained to the range [0,1].

Can produce negative probabilities.

Can produce probabilities greater than 1.

Relationship between probability and predictors is often
nonlinear.

Logistic Regression solves this using the sigmoid function,
ensuring 0 ≤ p ≤ 1

Logistic Regression

Logistic Regression is a supervised
machine learning algorithm used for
classification tasks. It predicts the
probability that an observation
belongs to a specific class (e.g.,
True/False, 0/1). Instead of fitting a
straight line like linear regression, it
uses a sigmoid function to create an Sshaped curve. [1, 2]
