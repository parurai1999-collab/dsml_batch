**Checking a Grade for students**

```python
grade = float(input("Enter the grade:"))
if grade > 4:
  print("Invalid grade")
elif grade >= 3.6 and grade <= 4:
  print("Grade A+")
elif grade >= 3.2 and grade <3.6:
  print("Grade A")
elif grade >= 2.8 and grade <3.2:
  print("Grade B+")
elif grade >= 2.4 and grade < 2.8:
  print("Grade B")
elif grade >= 2.0 and grade < 2.4:
  print("Grade C+")
elif grade >= 1.6 and grade < 2.0:
  print("Grade C")
elif grade >= 1.2 and grade <1.6:
  print("GradeD")
else:
  print("Grade f")
```

**Class Task For membership**

```python
age = int(input("Enter the age:"))
membership = input("Do you have membership:(yes/no)")
if membership == "yes" and age >= 40:
    print("60% Discount")
elif membership != "yes" and age >= 40:
    print("40%v discount")
else:
      print("No discount")
```

**If number is even then add the number and if odd then pop the item**

```python
lst = [3,4,6,7,9]
num = int(input("Enter the number:"))
if num % 2 == 0:
  lst.append(num)
  print(lst)
else:
  lst.pop()
  print(lst)
```
