Function Recursion



Find Factorial using Function recursion.

```python
def factorial(n):
  if n == 0:
    return 1
  return n * factorial(n-1)
print(factorial(5))
```

Write a recursive Python function fibonacci(n) that returns the nth Fibonacci number.


```python
def fib(n):
    if n <= 1:
      return n
    return fib(n-1) + fib(n-2)
nterms = 10
for i in range(10):

  print(fib(i))
```

Nested Function

```python
def outer(x):
  def inner(y):
    return x+y
  return inner
```

```python
add_ = outer(2)
res = add_(3)
print(res)
```

Pass Function As Argument

```python
def add(x,y):
  return x+y
def calculate(func, x,y):
  return func(x,y)
result = calculate(add,4,6)
print(result)
```

```python
def square(x):
  return x**2
def calculate(func,x):
  return func(x)
result = calculate(square,4)
print(result)
```

Return a function as Value Pass

```python
def greeting(name):
  def hello():
    return "Hello," +name+ "!"
  return hello
greet = greeting("Paru")
print(greet())
```

Decorators in python


```python
def myWrapper(func):
  def myFunc():
    print("Inner Wrapper")
    func()
  return myFunc
```

inner message haru display garna decorator is used

```python
@myWrapper
def myFunc():
  print("hello world")
myFunc()
```

```python
def myWrapper(func):
  def myFunc(*args, **kwargs):
    print(f"Calling {func.__name__}")
    return func(*args, **kwargs)
  return myFunc

```

```python
@myWrapper
def greet(name, age):
  print(f"Hello {name}, Age: {age}")
greet("paru", 23)
```

```python
def mywrapper(func):
  def myFunc(*args):
    print("Before function execution")
    result = func(*args)
    print("After function execution")
    return result
  return myFunc
```

```python
@mywrapper
def greet(name):
  print(f"Hello {name}")
greet("paru")
```
