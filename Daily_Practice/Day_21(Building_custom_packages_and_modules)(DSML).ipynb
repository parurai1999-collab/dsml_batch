**Python** **Modules** **and** **Packages**

Python uses modules and packages to organize code into reusable, maintainable
components.

**Module**

A module is a single Python
file (.py) containing code
such as functions, classes,
and variables.

**Package**

A package is a directory that
groups related modules
together

 **Install** **FastAPI**  **+** **Uvicorn**

uv add "fastapi"

uv add "uvicorn[standard]"


Built RESTful APIs using FastAPI in Visual Studio Code.

```python
from fastapi import FastAPI

app = FastAPI()

@app.get("/add")
def add_numbers(a: int, b: int):
    return {"result": a + b}
```

Manually ran the application server using Uvicorn for local deployment and testing.

uvicorn main:app --host 0.0.0.0 --port 8000

Accessed Swagger UI in the browser, and tested the endpoints to verify correct functionality using both query parameters and JSON input.
