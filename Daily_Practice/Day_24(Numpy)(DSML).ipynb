what is rank?

Number of dimensions in an array is called rank.

```python
#Rank 0 (Scalar)
a = np.array(10)
a.ndim
```

```python
#Rank 1 (vector) #Rank1 is called vector
a = np.array([1,2,3,4])
a.ndim
```

```python
# Rank 2 (Matrix) Rank2 is called matrix
a = np.array([[1,2,3], [4,5,6]])
a.ndim # ndim gives rank
```

```python
a.shape # It gives rows and column
```

Create 5*5 matrix and check the rank

```python
a = np.array([
    [[1,2,3,4,5], [2,3,4,5,6]],
    [[23,5,6,2,7],[4,6,7,8,0]],
    [[5,6,7,8,9],[1,2,7,58,8]],
    [[5,7,8,9,4],[3,52,7,8,9]],
    [[2,4,5,6,7],[2,1,5,6,7]]
    ])
a.ndim
a.shape
```

**np**.**zeros**() **and** **np**.**ones**()

np.zeros() and np.ones() are NumPy functions used to
create new arrays filled entirely with zeros (0) or ones (1).

```python
zeros = np.zeros((3,4))
print("Zero array: \n", zeros)
```

```python
ones = np.ones((4,5))
print("Zero array: \n", ones)
```

full() - creates an array with all elements initialized to same user-specified value.

```python
import numpy as np

arr = np.full((4, 5), fill_value=7)
print(arr)
```

```python
full = np.full((4,5), 9)
print(full)
```

arange() - create sequences of numbers (like Python’s range() but more powerful).


```python
arr = np.arange(2,10,3)
arr
```

np.linspace() - create evenly spaced numbers over a specified interval.

```python
import numpy as np

arr = np.linspace(0, 10, 5, dtype = int)
print(arr)
```

np.diag() - create a diagonal matrix or extract diagonal elements from a matrix.

```python
#diag() - create diagonal matrix
arr = np.diag([1,2,3])
arr
```

np.eye() - create an identity matrix (a matrix with 1s on the main diagonal and 0s elsewhere).

```python
#eye() - create matrix with ones on diagonal
arr = np.eye(3)
print(arr)
```

identity() - create an identity matrix

```python
# identity() - create identity matrix
arr = np.identity(3)
print(arr)
```

Difference between eye() and identity().

np.identity() → creates only a square identity matrix

np.eye() → more flexible, can create rectangular matrices and shifted diagonals

Identity matrix is always square and eye() can be implemented in rectangle

```python
# Always have to be square matrix. That's way it shows error.
arr = np.identity(3,5)
print(arr)
```
