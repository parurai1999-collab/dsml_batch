**Encapsulation**
**practice**

Create a BankAccount class with:

A public account holder name.

A protected account type.

A private PIN and balance.

Create a SavingAccount class that adds 5% interest to the
balance using methods.

```python
class BankAccount:
    def __init__(self, account_name, account_type, PIN, balance):
        self.account_name = account_name
        self._account_type = account_type
        self.__PIN = PIN
        self.__balance = balance

    def get_balance(self):
        return self.__balance


class Saving(BankAccount):

    def __init__(self, account_name, account_type, PIN, balance):
        super().__init__(account_name, account_type, PIN, balance)

    def add_interest(self):
        total_amount = 0.05 * self.get_balance()
        return total_amount


obj = Saving("Paru Rai", "Saving", 1234, 10000)
print(obj.add_interest())
```

**Abstraction**
**practice**

Create an abstract BankAccount class with abstract
methods deposit() and withdraw(). Implement them in a
SavingAccount class.

```python
from abc import ABC, abstractmethod

class BankAccount(ABC):
    @abstractmethod
    def deposit(self):
        pass

    def withdraw(self):
        pass


class SavingAccount(BankAccount):
    def deposit(self):
        return "Amount deposit"

    def withdraw(self):
        return "Amount withdraw"


d1 = SavingAccount()
print(d1.deposit())
w1 = SavingAccount()
print(w1.withdraw())
```

```python
from abc import ABC, abstractmethod


class BankAccount(ABC):
    @abstractmethod
    def deposit(self, method):
        pass

    def withdraw(self, amount):
        pass


class SavingAccount(BankAccount):
    def __init__(self, balance=0):
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        print(f"Deposit ${amount}")

    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
            print(f"Withdraw ${amount}")
        else:
            print("Insufficient Balance")

    def check_balance(self):
        print(f"Balance: ${self.balance}")


account = SavingAccount(1000)
account.deposit(500)
account.withdraw(200)
account.check_balance()
```

```python
class PaymentGateway:
  def process_payment(self, amount):
    pass
class CreditCard(PaymentGateway):
    def process_payment(self, amount):
      return f"Processing ${amount} via Credit Card Authentication."
class PayPal(PaymentGateway):
  def process_payment(self, amount):
    return f" Processing ${amount} via Paypal secure token."
def execute_checkout(payment_method, amount):
  print(payment_method.process_payment(amount))
card = CreditCard()
wallet = PayPal()
execute_checkout(card, 250)
execute_checkout(wallet, 250)
```

```python
class Student:
  def __init__(self, m1, m2):
    self.m1 = m1
    self.m2 = m2
  def __add__(self, other):  #student.__add__(s1,s2)
    m1 = self.m1 + other.m1
    m2 = self.m2 + other.m2
    s3 = Student(m1,m2)
    return s3
  def __gt__(self, other):
    m1 = self.m1 + other.m2
    m2 = self.m2 + other.m2
    if m1 > m2:
      return True
    else:
      return False
```

```python
s1 = Student(56, 98)
s2 = Student(90, 87)
s3 = s1 + s2
```
