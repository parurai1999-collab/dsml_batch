**Metrics**

Evaluation metrics are quantitative measures used to assess
the performance of machine learning models.


**Confusion** **Matrix**


A confusion matrix compares Actual values with Predicted values.


TP: Model correctly predicts Positive.

TN: Model correctly predicts Negative.

FP: Model predicts Positive but actually Negative.

FN: Model predicts Negative but actually Positive.

**Accuracy**

Accuracy measures the overall correctness of the model.

Accuracy=TP+TN/ TP + TN + FP + FN

**Precision**

Precision tells us:

Out of all predicted positives, how many were actually positive?

Precision=TP/ TP + FP

**Recall** (**Sensitivity** / **True** **Positive** **Rate**)

Recall tells us:

Out of all actual positives, how many did the model correctly identify?

Recall=TP/ TP + FN

**F1**-**Score**

F1-score is the harmonic mean of Precision and Recall.

F1=2 × (Precision * Recall / Precision + Recall​)

```python
tp = 50
fn = 10
fp = 15
tn = 25
def accuracy(tp, tn, fp, fn):
  return (tp + tn) / (tp + tn + fp + fn)
def precision(tp, fp):
  return tp / (tp + fp)
def recall(tp, fn):
  return tp / (tp + fn)

def f1_score(tp, fp, fn):
  p = precision( tp, fp)
  r = recall(tp, fn)
  return 2 * (p*r) / (p+r)


print("Accuracy:", accuracy(tp, tn, fp, fn))
print("Precision:", precision(tp, fn))
print("Recall:", recall(tp, fn))
print("F1 Score:", f1_score(tp, fp, fn))
```

**Underfitting**

Underfitting occurs when a model is too simple to capture the underlying patterns in the data.

Characteristics
Poor performance on training data.
Poor performance on testing data.
High bias, low variance.

**Overfitting**

Overfitting occurs when a model learns not only the actual patterns but also the noise in the training data.

Characteristics
Very high performance on training data.
Poor performance on testing data.
Low bias, high variance

**Regularization** is one of the most effective techniques to deal with overfitting.

What is Regularization?

Regularization adds a penalty term to the loss (cost) function to discourage the model from learning overly complex patterns.

It helps:

Reduce overfitting
Improve generalization
Prevent very large coefficient values

**Types** **of** **Regularization**

L1 Regularization (Lasso)

Adds the absolute value of coefficients as a penalty.


Features:

Can make some coefficients exactly 0.

Performs feature selection.


2. L2 Regularization (Ridge)

Adds the squared value of coefficients as a penalty.


Features:

Shrinks coefficients toward zero.

Does not make them exactly zero.

Most commonly used.
