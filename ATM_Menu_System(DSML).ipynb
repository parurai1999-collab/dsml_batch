**ATM Menu System**

```python
balance = 1000
menu = {
    "Option 1": "Check_Balance",
    "Option 2": "Deposit",
    "Option 3": "Exit"
}
while True:

    print(f"""
    ---------------------------
      Option 1: {menu["Option 1"]}
      Option 2: {menu["Option 2"]}
      Option 3: {menu["Option 3"]}
    ---------------------------
  """)

    choice = int(input("Enter the option:(1/2/3)"))
    if choice == 1:
        print("Your balance is ", balance)
    elif choice == 2:
        amount = float(input("Enter the amount:"))
        balance += amount
        print(f'{amount} deposited successfylly.')
        print(f'New balance: {balance}')

    elif choice == 3:
        print("Thank you visiting ATM.")
        break
    else:
        print("Invalid Entry")
```
