**filter**() **in** **Python**


```python
lst = [1,2,3,4,5,6,7,8,9]
def is_even(x):
  return x % 2 == 0
result = filter(is_even, lst)
print(list(result))
```

**reduce**() **in** **Python**

**Need** **a** **single** **value** **then** **we** **use** **reduce**()

```python
from functools import reduce
def mux(a,b):
  return(a*b)
lst = [1,2,3,4,5]
result = reduce(mux, lst)
print(result)
```

**Find** **maximum** **number** **using** **reduce**() **function**.

```python
from functools import reduce
def max(a,b):
  if a > b:
    return a
  else:
    return b
lst1 = [1, 6, 7, 9, 4]
result = reduce(max, lst1)
print(result)
```

**kwargs

```python
data2 = {
    "name" : "paru",
    "age" : 23
}
def joined2(**kargs):
  return kargs
  print(joined2(data2))
```

```python
def data3(n, *args, **kwargs):
  print(n)
  print(args)
  print(kwargs)
data3(3,2,3,4,5 , s1=1, s2=4)
```
