**NumPy** - **Element**-**wise** **Matrix** **Operations**


**Matrix** **Addition**

```python
import numpy as np
arr1 = np.array([[1,2],[3,6]])
arr2 = np.array([[1,2], [3,6]])
addition = np.add(arr1, arr2)
print(addition)
```

**Matrix** **Subtraction**

```python
subtraction = np.subtract(arr1, arr2)
print(subtraction)
```

**Matrix** **Multiplication**

```python
multiplication = np.multiply(arr1,arr2)
print(multiplication)
```

```python
import numpy as np

A = np.array([
    [1,  1, 2, 1, 1],
    [2, -1, 1, 3, 2],
    [3,  1, 2, 2, -1],
    [1, -1, 1, 2, 3],
    [1,  2, 3, 1, 1]
])


E = np.array([5, 8, 4, 10, 7])

x = np.linalg.solve(A, E)

print(x)
```

```python
import numpy as np

X = np.array([
    [1, 1],
    [2, 1],
    [3, 1]
])

Y = np.array([2, 3, 5])

beta = np.linalg.inv(X.T @ X) @ X.T @ Y

print(beta)
print("Slope:", beta[0])
print("Intercept:", beta[1])
print("Intercept:", beta[1])
```

```python
import numpy as np

X = np.array([
    [1, 1],
    [2, 1],
    [3, 1]
])

Y = np.array([2, 3, 5])

beta = np.linalg.inv(X.T @ X) @ X.T @ Y

print(beta)
print("Slope:", beta[0])
print("Intercept:", beta[1])
```

**Determinants** **&** **Inverse**

```python
import numpy as np
A = np.array([[1,2],
              [4,5]])
det = np.linalg.det(A)
print("Determinant:", det)
```

```python
inv = np.linalg.inv(A)
print("Inverse Matrix:", inv)
```

Eigenvalues and Eigenvectors

```python
eigenvalues, eigenvectors = np.linalg.eig(A)
print("Eigen Valies:", eigenvalues)
print("Eigen Vector:", eigenvectors)
```

Singular Value
Decomposition (SVD)

```python
U,S,V = np.linalg.svd(A)
print("U matrix: \n", U)
print("Sigma values:", S)
print("V matrix:", V)
```
