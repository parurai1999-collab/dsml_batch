**Python** **Property** **Decorator**

The @property decorator in Python is used to define a method that can be accessed like an attribute (without calling it like a function).

```python
class Student:
  def __init__(self, name):
    self.name = name

  @property
  def name(self):
    return self._name
```

```python
s = Student("Paru")
s.name
```

getter (@property)

A getter is called when you read a property.


```python
class Student:
  def __init__(self, name):
    self._name = name

  @property
  def name(self):
    print("Getter is called")
    return self._name
```

```python
s = Student("Ram")
print(s.name)
```

Setter (@property)

A setter is called when you assign a value.

```python
class Student:
  def __init__(self, age):
    self._age = age
  @property
  def age(self):
    return self._age
  @age.setter
  def age(self, value):
     print("Setter called")
     if value < 0:
       raise ValueError("Age cannot be negative")
     self._age = value
```

```python
s =Student(20)
s.age
```

```python
s.age = 40
s.age
```

Deleter (@deleter)

A deleter is called when you use del.

Purpose

Clean up resources

Remove sensitive data

Control deletion behavio


```python
class Student:
  def __init__(self, age):
    self._age = age
  @property
  def age(self):
    return self._age
  @age.deleter
  def age(self):
    print("Deleting age ...")
    del self._age
```

```python
s = Student(20)
del s.age
```

```python
class Student:
    def __init__(self, name, age):
        self._name = name
        self._age = age

    @property
    def name_age(self):
        return f"{self._name} is {self._age} years old"

    @name_age.setter
    def name_age(self, value):
        print("Setter called")

        name, age = value   # unpack tuple

        if not isinstance(name, str) or age < 0:
            raise ValueError("Invalid input")

        self._name = name
        self._age = age
```

```python
s = Student("Paru", 22)

print(s.name_age)


s.name_age = ("Ram", 25)

print(s.name_age)
```

Validating input (email, phone number, password)

Searching text

Extracting data from strings

Replacing text
