**Check if number is positive and integer**

```python
number = int(input("Enter a number:"))
if number % 2 == 0 and number > 0:
  print("Number is positive and even number.")
else:
    print("Number is not postive and even number")
```

**Check** **if** **a** **user** **is** **not** **logged** **in**.


```python
name = "paru"
user_name = input('Enter the username:')
if user_name != name:
  print('User is not logged in.')
else:
  print("User is logged in.")
```

**Check** **if** **age** **is** **between** **18** **and** **60** **using** **and**.


```python
age = int(input("Enter your age:"))
if age >= 18 and age <= 60:
  print("your age is between 18 and 60")
else:
  print("Your age is not between 18 and 60")
```

**Check** **if** **a** **day** **is** **Saturday** **or** **Sunday** **using** **or**.


```python

day = input("Enter the day:").lower()
valid_days= ['saturday', 'sunday', 'monday', 'tuesday', 'wednesday', 'thursday', 'friday']
if day != valid_days:
  print("Invalid entry")
elif day == "Saturday" or day.lower() == "Sunday":
  print("Today is weekend")
else:
  print("Today is weekday.")
```
