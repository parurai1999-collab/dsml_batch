**Lambda** **Function**

```python
def sum(a,b):
  return a+b
sum(2,4)
```

**Lambda is nameless or anonymous function that are quick and simple, and we use them for short tasks**

```python
data = lambda a,b : a+b
data(4,5)
```

```python
func = lambda n : n*n
func(6)
```

**Python** **lambda** **function**
**Using** **map**()

```python
num = [1,2,3,4,5,6]
list(map(lambda n : n*n,num))
```

```python
list(filter(lambda n : n*n,num))
```

**Python** **lambda** **function**
**Using** **filter**()

```python
num = [1,2,3,4,5,6]
list(filter(lambda n : n % 2 == 0, num))
```

**Python** **lambda** **function**
**Using** **reduce**()

```python
from functools import reduce
num = [1,2,3,4,5,6]
reduce(lambda a,b :a *b, num)
```

```python
num = int(input("Enter the number:"))
result = lambda num: "even" if num%2 == 0 else "odd"
result(num)
```

**Python** **lambda** **function**
**Sort** **Dictionary** **List**

```python
students = [
    ("Ram",80),
    ("Hari",95),
    ("Sita",85)
]
students.sort(key=lambda x:x[1])
print(students)
```

```python
students.sort(reverse=True, key=lambda x:x[1])
print(students)
```

**Python** **List** **comprehension**

[expression for item in iterable if condition]


```python
data = ["welcome", "to", "Dsml", "course"]
for word in data:
  if "e" in word:
    print(word.upper())
```

```python
[word.upper() for word in data if "e" in word]
```

```python
lst = [1,2,3,4,5,6]
["even" if num%2 == 0 else "odd" for num in lst]
```

```python

[i for i in range(10)]
```

```python
list1 = [1,2]
list2 = ["A", "B"]
pairs = [(x,y) for x in list1 for y in list2]
print(pairs)
```

**Python** **Dictionary** **Comprehension**

{key: value for item in iterable if condition}

```python
{num: num*num for num in range(1,11)}
```

```python
lst3 = [1,2,3,4,5,6,7,8,9]
{num: num%2==0 for num in lst3}
```

```python
{num:num**2 for num in lst3 if num%2 == 0}
```

```python
original_dict = {"ram":38, "sita":48, "gita":57, "john":33}
{(key,value) for key, value in original_dict.items() if value%2 == 0}
```

```python
original_dict = {"ram":38, "sita":48, "gita":57, "john":33}
new_dict = {key:value for key, value in original_dict.items() if value % 2 != 0 if value < 40}
print(new_dict)
```
